```python
import numpy as np 
import pandas as pd 
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("../data/telco_churn_training.csv")
df.shape
```

```python
df_train , temp = train_test_split(
    df,
    shuffle=True,
    random_state=1,
    stratify=df["Churn"],
    test_size=0.3
)
df_val , df_test= train_test_split(
    temp,
    shuffle=True,
    random_state=1,
    stratify=temp["Churn"],
    test_size=0.5
)
```

```python
x_train = df_train.drop(columns=["Churn", "customerID"])
x_val = df_val.drop(columns=["Churn", "customerID"])
x_test = df_test.drop(columns=["Churn", "customerID"])

y_train = df_train["Churn"]
y_val = df_val["Churn"]
y_test = df_test["Churn"]
df.dtypes
```

```python
num = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
]
categ = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]
```

```python
TotalCharges_pipline = Pipeline([
    ("imputer",SimpleImputer(strategy="constant", fill_value=0)),
    ("scala",StandardScaler())
])
preprocessor = ColumnTransformer([
    ("imputer",TotalCharges_pipline,["TotalCharges"]),
    ("cate",OneHotEncoder(handle_unknown="ignore"),categ),
    ("num",StandardScaler(),num),
    
])

First_model = Pipeline([
    ("Preprocessor",preprocessor),
    ("model",LogisticRegression())
])
```

```python
First_model.fit(x_train,y_train)
```

```python
y_pred = First_model.predict(x_val)
y_pred
```

```python
accuracy = accuracy_score(y_val, y_pred)

print(accuracy)
```

```python
Seconed_model = Pipeline([
    ("Preprocessor",preprocessor),
    ("model",DecisionTreeClassifier(random_state=1,max_depth=100,min_samples_split=10,min_samples_leaf=10))
])

```

```python
Seconed_model.fit(x_train,y_train)
```

```python
y_pred_2 = Seconed_model.predict(x_val)
accuracy_2 = accuracy_score(y_val, y_pred_2)

print(accuracy_2)
```

```python
Third_model = Pipeline([
    ("Preprocessor",preprocessor),
    ("model",RandomForestClassifier(random_state=1,max_depth=200,min_samples_split=10,min_samples_leaf=10))
])
```

```python
Third_model.fit(x_train,y_train)
```

```python
y_pred_3 = Third_model.predict(x_val)
accuracy_3 = accuracy_score(y_val, y_pred_3)

print(accuracy_3)
```
